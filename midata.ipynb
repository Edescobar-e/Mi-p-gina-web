```python
import zipfile
import os

with zipfile.ZipFile("archive.zip", "r") as zip_ref:
    zip_ref.extractall("/content/spotify_temp")

print("Archivo descomprimido correctamente")
```

```python
import os
import shutil

os.makedirs("/content/spotify-2023/data/raw", exist_ok=True)
os.makedirs("/content/spotify-2023/data/processed", exist_ok=True)

for root, dirs, files_list in os.walk("/content/spotify_temp"):
    for file in files_list:
        if file.endswith(".csv"):
            shutil.copy(
                os.path.join(root, file),
                "/content/spotify-2023/data/raw/spotify-2023.csv"
            )

print("Estructura creada correctamente")
```

```python
import pandas as pd

ruta = "/content/spotify-2023/data/raw/spotify-2023.csv"

df = pd.read_csv(ruta, encoding="latin1")

print("Dimensiones originales:", df.shape)
df.head()
```

```python
columnas_numericas = [
    "artist_count", "released_year", "released_month",
    "released_day", "in_spotify_playlists", "in_spotify_charts",
    "streams", "in_apple_playlists", "in_apple_charts",
    "in_deezer_playlists", "in_deezer_charts",
    "in_shazam_charts", "bpm", "danceability_%",
    "valence_%", "energy_%", "acousticness_%",
    "instrumentalness_%", "liveness_%", "speechiness_%"
]

for columna in columnas_numericas:
    df[columna] = pd.to_numeric(
        df[columna].astype("string").str.replace(",", "", regex=False),
        errors="coerce"
    )

df = df.dropna(subset=["streams"]).copy()

columnas_texto = ["track_name", "artist(s)_name", "key", "mode"]

for columna in columnas_texto:
    df[columna] = df[columna].astype("string").str.strip()

df = df.drop_duplicates()

print("Dimensiones después de la limpieza:", df.shape)
df.head()
```

```python
ruta_limpia = "/content/spotify-2023/data/processed/spotify-2023-clean.csv"

df.to_csv(ruta_limpia, index=False, encoding="utf-8")

print("Datos limpios guardados correctamente")
```

```python
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

print("\nValores faltantes:")
print(df.isnull().sum())

print("\nDuplicados:")
print(df.duplicated().sum())

print("\nTipos de datos:")
print(df.dtypes)
```
